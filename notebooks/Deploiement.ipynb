```python
from google.colab import drive
import shutil
import os

# Mount Google Drive
drive.mount('/content/drive')

# Paths
drive_model_output_path = '/content/drive/My Drive/model_output/'
colab_model_output_path = '/content/model_output/'

# Check if the folder exists in Google Drive
if os.path.exists(drive_model_output_path):
    # Copy the folder to Colab's working directory
    if not os.path.exists(colab_model_output_path):
        shutil.copytree(drive_model_output_path, colab_model_output_path)
        print(f"Folder copied to: {colab_model_output_path}")
    else:
        print(f"Folder already exists in Colab at: {colab_model_output_path}")
else:
    print(f"The folder '{drive_model_output_path}' does not exist in Google Drive. Please check the path.")
```
